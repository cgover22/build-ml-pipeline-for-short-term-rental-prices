## Title: Exploratory Data Analysis (EDA)
## Purpose: Gain meaningful insights on the data to share with stakeholders. Come to an agreement on what actions should be taken on the data before passing it over for data cleaning step in the ML Pipeline.

### Step: Fetch the artifact sample.csv from Weights & Biases and read it into a pandas dataframe. Use code=True so the notebook is uploaded and versioned in W&B

```python
import wandb
import pandas as pd

run = wandb.init(project="nyc_airbnb", group="eda", save_code=True)
local_path = wandb.use_artifact("sample.csv:latest").file()
df = pd.read_csv(local_path)
```

### Step: Generate the Profile Report using ydata-profiling

```python
import ydata_profiling

profile = ydata_profiling.ProfileReport(df)
profile.to_notebook_iframe()
```

### Step: Apply fixes. After reviewing the profile report and talking it over with stakeholders, we will restrict prices to between 10 dollars and 350 dollars. We will also convert the last_review column from string to datetime.

```python
# Drop outliers
min_price = 10
max_price = 350
idx = df['price'].between(min_price, max_price)
df = df[idx].copy()

# Convert last_review to datetime
df['last_review'] = pd.to_datetime(df['last_review'])
```

### Step: Verify the fixes.

```python
df.info()
```

### Step: Verify the fixes. Confirm prices are within the defined range

```python
print('price range')
mins = df['price'].min()
maxs = df['price'].max()
print(f'The min value is: {mins}')
print(f'The max value is: {maxs}')
```
